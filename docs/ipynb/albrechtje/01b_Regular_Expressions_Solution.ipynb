## Regular expressions

Regular expressions are a powerful tool for searching texts for certain patterns and extracting or modifying them if necessary. All programming languages offer options for working with regular expressions. Especially in scripting languages such as Python, Perl or shell programming (e.g. Bash), regular expressions are often used.

Regular expressions are used for the following purposes, for example:
- Checking whether entries in HTML forms correspond to the specifications (e.g. zip code or email address)
- Searching for the occurrence of certain patterns in texts (e.g. extracting links from `href` tags on a web page)
- Extraction of certain parts of text (e.g. only the area code of a telephone number)
- Replacing certain parts of text (e.g. anonymization of secret numbers by "\*\*\*\*")
- Splitting larger texts into smaller parts (e.g. into paragraphs or words)

Python has the standard library `re` for regular expressions and the newer, backwardcompatible library `regex` that offers support for POSIX character classes, unicode and some more flexibility. You should import it like this:


```python
import regex as re
```

### Structure of regular expressions

The following example shows how numbers can be extracted from a string using a simple regular expression. The regular expression `[0-9]+` describes the search pattern.

```python
import regex as re
numbers = re.findall(r'[0-9]+', '90489 Nürnberg, Kesslerplatz 12')
print(numbers)
```

A single character is checked for a digit using the expression ``[0-9]``, which is equivalent to ``[01234567879]``. In order to match sequences of characters, there are various quantifiers that can be placed after a simple expression. The following table gives an overview of the available quantifiers and their meaning.


  | Quantor | Description |
 |--------------------------|------------------------------------|
 | ``?`` | none or exactly one occurrence of the preceding character |
 | ``*`` | zero or more repetitions |
 | ``+`` | One or more repetitions |
 | ``{n,m}`` | Between n and m repetitions |
 | ``{n}`` | Exactly n repetitions |

A few examples illustrate the handling:

  | Expression | Examples of matches | Description
 |----------|---------------|----------------
 | [abc] | 'a', 'b', 'c' | one character: a, b, or c
 | (ab&#124;cd) | 'ab' or 'cd'| &#124; denotes "or" of character strings
 | [a-z] | 'a', 'b', ..., 'z' | lowercase letter - be careful with umlauts!
 | [a-zäöüß] | 'a', 'z', 'ö', 'ü' | Lowercase letter including umlauts
 | [A-Za-z]\* | 'hello', 'why', 'aaBBcc', '' | Zero or more repetitions of upper or lower case letters
 | [0-9]+ | '1', '2434', '76553' | one or more repetitions of digits
 | [0-9]?   | '', '0', '5' | none or one digit
 | [0-9]{4} | '3415', '5423' | four-digit sequence of digits
 | .* | '', 'so that', 'hello123' | Any character string (may also be empty)
| I .+ [Ll]ove | 'I absolutely Love', 'I secretly love' | 'I', any non-empty sequence of characters, 'Love' or 'love'
| D[^o]g | 'Dig', but not 'Dog' | [^o] stands for "not o"

Predefined character classes exist in the context of regular expressions. Instead of individual characters, these define entire search patterns. The following table provides an overview of frequently used classes.

  | Operator | Description |
 |-----------|------------------------------------------------------------------|
 | ``.`` | Any letter except a newline (\n) |
 | ``\d`` | Any decimal number (equivalent to [0-9]) |
 | ``\D`` | Any character that is not a number (equivalent to [^0-9]) |
 | ``\s`` | Any space |
 | ``\S`` | Any character that is not a space |
 | ``\w`` | Any alphanumeric character (equivalent to [a-zA-Z0-9_]) |
 | ``\W`` | Any non-alphanumeric character (equivalent to [^a-zA-Z0-9_])|
 | ``[...]`` | List of allowed characters (matches one character) |
 | ``^`` | NOT (when used within a listing) |
 | ``\/`` | Matches the character / |
 | ``\`` | Matches the character \ |
 | ``^ and $`` | ^ and $ match the start and end of a string |
 | ``\t, \n, \r`` | Matches tab, new line and return

### Regular expressions in Python

The ``re`` module (https://docs.python.org/3/library/re.html) offers several methods that can be used to search strings for regular expressions:

- ``re.match()``: Search at the beginning of the string
- ``re.search()``: search the entire string
- ``re.findall()``: Output of all hits in the string
- ``re.sub()``: Replace regular expressions
- ``re.split()``: Splitting using regular expressions
- ``re.compile()``: Create reusable pattern objects

The first four functions are briefly discussed here.

Details here: https://docs.python.org/3/howto/regex.html#the-backslash-plague

**Raw strings**

To deal with the special meanings of backslashes in particular, regular expressions should be specified as raw strings, preceded by an `r`. The behavior is similar to the '@' character in C# strings: Backslashes are not treated specially in the string:

```python
s1 = '\n'
print(s1, len(s1))

s2 = r'\n'
print(s2, len(s2))
```

#### re.match(pattern, string)

The method searches for a suitable result at the beginning of the string.

```python
import re
match = re.match(r'The.*sits', 'The cat sits on the roof.')
```

The returned [Match object](https://docs.python.org/3/library/re.html#match-objects) contains the matching regular expression:

```python
match.group(0)
```

Parts of the matched expression (groups) can be output using bracket expressions. `group(0)` always contains the entire expression.

```python
match = re.match(r'The (.*) sits on the (.*)\.', 'The cat sits on the roof.')

match.groups()
match.group(0)
match.group(1)
match.group(2)
```

#### re.search(pattern, string)
``search()`` is a bit more general than ``match()``, since results are not limited to the beginning of the search string, but otherwise works the same way:

```python
match = re.search(r'cat \w+', 'The cat sits on the roof.')
print(match)
```

```python
match.group(0)
```

####  re.findall (pattern, string):
The ``findall()`` method returns a list of all results matching the regular expression. There are no restrictions regarding the start and end of the search.

```python
import re

digits = re.findall(r'[0-9]', '90489 Nürnberg, Kesslerplatz 12')
print(digits)

numbers = re.findall(r'[0-9]+', '90489 Nürnberg, Kesslerplatz 12')
print(numbers)

# Extracting domain names from emails
domains = re.findall(r'@\w+.([\w\.]+)', 'abc.test@gmail.com, xyz@web.de, asdf@stuff.co.uk')
print(domains)

# Finding occurrences of the word 'car' in its singular and plural forms
car = re.findall(r'[Cc]ars?', 'The car was parked in the parking lot automatically among other cars.')
print(car)
```

With ``sub()`` (Substitute) no regular expression can be replaced in a string. The method generalizes the replace function for strings.

```python
result = re.sub(r'[0-9]+', '***', 'The secret code is: 4711')
print(result)
```

#### Flags

The operation of the `re` methods can be controlled via flags. Two examples are shown here.

Details here: https://docs.python.org/3/howto/regex.html#compilation-flags

```python
text = """
There once was a frisbee, quite grand;
It flew too far, over a land.
The frisbee soared away,
And didn't come back all day.
The crowd - for hours on end -
Waited for its return, my friend.
"""

# Finds nothing
re.findall(r'soared.*day', text)
# re.DOTALL allows the dot to match across newline characters
re.findall(r'soared.*day', text, flags=re.DOTALL)

# Finds nothing, as "the" is capitalized
re.findall(r'the', text)
# re.IGNORECASE ignores case
re.findall(r'the\w*', text, flags=re.IGNORECASE)
```

#### Greedy and Reluctant Behavior

Assuming that the two greetings ``Hello``and ``Salut`` are to be extracted from the string ``English: "Hello", French: "Salut"``. One might assume that the search can be performed relatively easily with the expression ``"(.+)"`` (opening quotation mark, searched string, closing quotation mark). However, this is not the case. In fact, `findall` returns the following result:

```python
text = 'english: "Hello", french: "Salut"'

greetings = re.findall(r'"(.+)"', text)
print(greetings)
```

The reason for this is that the quantor is based on a so-called "greedy behavior" by default. The quantor tries to match as much as possible and deliver the best possible search result. It therefore matches everything up to the last quotation mark (the one after "Salut").

To find the smallest possible match ("Reluctant Behavior"), a ``?`` must be appended to the quantifier (for example ``*?`` or ``+?``):

```python
greetings = re.findall(r'"(.+?)"', text)
print(greetings)
```
