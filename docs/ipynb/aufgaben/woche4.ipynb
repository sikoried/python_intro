# Vollkommene Zahlen

Schreiben Sie ein Programm, welches zeilenweise Ganzzahlen einliest, und für jede `true` ausgibt, wenn sie eine [vollkommene Zahl](https://de.wikipedia.org/wiki/Vollkommene_Zahl) ist, oder `false`, wenn nicht.
**Hinweise:** 
- Es kommen Zahlen größer als `2147483647` aber kleiner als `9223372036854775807` vor.
- Wenn sie die Herausforderung suchen, wählen Sie nicht die allerleichteste (und allerschnellste) Lösung.

# Beispiele

## Eingabe
```
6
28
496
1000
8128
10000
```

## Ausgabe
```
true
true
true
false
true
false
```


````python
beispiele = [
    (6, True),
    (28, True),
    (496, True),
    (1000, False),
    (8128, True),
    (10000, False),
    (2305843008139952128, True)
]

def vollkommen(z: int):
    s = 1
    for t in range(2, int(z/2)+1):
        if z % t == 0:
            s += t
    return z == s

for z, res in beispiele:
    print(z, res, vollkommen(z))
    assert vollkommen(z) == res
````

# Klammersalat

Überprüfen Sie für jede Zeile der Eingabe, ob die Anzahl der Klammern korrekt ist, d.h. dass alle öffnenden Klammern auch geschlossen werden, und geben Sie entsprechend `OK` aus wenn alles in Ordnung ist.
Stimmt die Klammerung nicht, sollen Sie ausgeben
- an welcher Stelle (von 0 gezählt) der erste Fehler auftritt, 
- welches Klammer gefunden wurde (oder `-` am Ende der Zeile),
- und welches Klammer stattdessen erwartet wäre (bzw. `-` wenn keine).

**Beachten Sie:** In dieser Version der Aufgabe sind alle ASCII Klammersymbole, also `()`, `[]`, `{}` und `<>`, erlaubt.
Achten Sie insbesondere darauf, dass die jeweilige schließende Klammer vom gleichen Symbol sein muss, d.h. Klammerpaare können sich nicht überlappen (siehe Beispiele unten), sondern nur eingebettet sein.


# Beispiele:

## Eingabe
```
(Hallo, Welt)
(das ist falsch}
(das { ist auch ) falsch)
das (ist { ein < langes ( aber ) richtiges > [ Beispiel ] } ! )
)
{
(a(b(c)
hans)
```

## Ausgabe
```
OK
15 } )
16 ) }
OK
0 ) -
1 - }
7 - )
4 ) -
```


````python
beispiele = [
    ("(Hallo, Welt)", True),
    ("(das ist falsch}", (15, '}', ')')),
    ("(das { ist auch ) falsch)", (16, ')', '}')),
    ("das (ist { ein < langes ( aber ) richtiges > [ Beispiel ] } ! )", True),
    (")", (0, ')', '-')),
    ("{", (1, '-', '}')), 
    ("(a(b(c)", (7, '-', ')')),
    ("hans)", (4, ')', '-'))
]

def validate(a):
    klammern = list()
    paare = {
        '(': ')',
        ')': '(',
        '{': '}',
        '}': '{',
        '[': ']',
        ']': '[',
        '<': '>',
        '>': '<',
    }
    for i, z in enumerate(a):
        if z in '({[<':
            klammern.append(z)
        elif z in ')]}>':
            if len(klammern) == 0:
                return (i, z, '-')  # problem: Klammer, aber keine offen
            elif paare[z] == klammern[-1]:
                del klammern[-1]
            else:
                return (i, z, paare[klammern[-1]]) # problem: falsche Klammer
    if len(klammern) == 0:
        return True
    else:
        return (len(a), '-', paare[klammern[-1]])
            

for a, res in beispiele:
    print(a, res, validate(a))
    assert validate(a) == res
````

# Buchstabieren

Schreiben Sie ein Programm, das Zeilen mit jeweils zwei Wörtern einliest, die dann verglichen werden sollen.
Geben Sie jeweils aus, ob diese gleich sind (`richtig`) oder sich in höchstens einem Buchstaben unterscheiden (`ok`), ansonsten `falsch`.

# Beispiele

# Eingabe
```
Sample Simple
Sample simple
Sample Sample
Haus Hans
Hauswein Hanswurscht
wein ein
```

Ausgabe:
```
ok
ok
richtig
ok
falsch
ok
```

Bonusaufgabe: Lösen Sie die Aufgabe so, dass eine fest Anzahl Unterschiede größer 1 möglich ist.

````python
beispiele = [
    ("Sample Simple", "ok"),
    ("Sample simple", "ok"),
    ("Sample Sample", "richtig"),
    ("Haus Hans", "ok"),
    ("Hauswein Hanswurscht", "falsch"),
    ("wein ein", "ok")
]

def pruefe(s: str):
    w1, w2 = s.lower().split()
    if len(w1) == len(w2):
        # Wörter gleich?
        if w1 == w2:
            return 'richtig'
        
        # prüfe, ob höchstens ein Unterschied
        anz = 0
        for i in range(len(w1)):
            if w1[i] != w2[i]:
                anz += 1
        if anz > 1:
            return 'falsch'
        else:
            return 'ok'
    elif abs(len(w1) - len(w2)) == 1:
        # prüfe, ob kürzeres in längerem enthalten
        if len(w1) > len(w2):
            return 'ok' if w2 in w1 else 'falsch'
        else:
            return 'ok' if w1 in w2 else 'falsch'
    else:
        return 'falsch'

for s, res in beispiele:
    print(s, res, pruefe(s))
    assert pruefe(s) == res
````
