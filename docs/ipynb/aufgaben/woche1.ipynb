# Diamantendrucker

Dieses Programm soll einen Diamanten, bestehend aus Sternen (`*`) ausgeben. 
Die Eingabe ist eine positive Zahl größer als zwei, welche die Anzahl der Sternsymbole auf der Mittellinie angibt.
Die Reihen darüber und darunter sollen jeweils immer zwei Sterne weniger besitzen als die Mittellinie.

# Beispiel:

Eingabe:
```python
n = 5
```

Ausgabe:
```
  *
 ***
*****
 ***
  *
```


Eingabe:
```python
n = 4
```

Ausgabe:
```
 **
****
 **
```


````python
n = 11

anz_leer = int((n-1)/2)
anz_stern = 2 - (n % 2)
letzte = int(n / 2) - (1 - n % 2)

print(letzte)

zeilen = []

for x in range(letzte + 1):
    zeilen.append('-' * anz_leer + '*' * anz_stern)
    anz_stern = anz_stern + 2
    anz_leer = anz_leer - 1

zeilen = zeilen + list(reversed(zeilen[:-1]))

for z in zeilen:
    print(z)
````

# Berechnete Determinanten

Schreiben Sie ein Programm, welches die Determinanten von 2x2 und 3x3 Matrizen berechnet

# Beispiele:

### 1. 2x2 

Eingabe:
```python
mat = [[2, 2], [2, 2]]
```

Ausgabe:
```python
0
```

### 2. 3x3

Eingabe:
```python
mat = [[2, 5, 2], [3, -3, 1], [1, 4, -4]]
```

Ausgabe:
```
111
```


````python
mat1 = [[2, 2], [2, 2]]
mat2 = [[2, 5, 2], [3, -3, 1], [1, 4, -4]]

det1 = mat1[0][0] * mat1[1][1] - mat1[1][0] * mat1[0][1]

print(det1)

det2 = mat2[0][0] * (mat2[1][1]*mat2[2][2] - mat2[1][2]*mat2[2][1]) \
    - mat2[1][0] * (mat2[0][1]*mat2[2][2] - mat2[0][2]*mat2[2][1]) \
    + mat2[2][0] * (mat2[0][1]*mat2[1][2] - mat2[1][1]*mat2[0][2])

print(det2)
````

# Scheckheft

Schreiben Sie ein Programm, dass zeilenweise eingelesene Dezimalzahlen jeweils als Ganzzahl gefolgt von einem "gekürztem" Bruch ausgibt.
Hinweis: Berücksichtigen Sie sowohl `,` als auch `.` als Punktuation.

# Beispiele:


| Eingabe | Ausgabe |
|:-------:|:-------:|
| 2.95    | 2 19/20 |
| 3,4     | 3 2/5   |
| 1       | 1       |



````python
def brueche(zahl1):
    hilfe = zahl1
    zahl1 = ''

    for b in hilfe:
        if b == ',':
            zahl1 = zahl1 + '.'
        else:
            zahl1 = zahl1 + b

    z = float(str(zahl1))

    ganzzahl = int(z)

    if '.' not in zahl1:
        return ganzzahl

    ganzzahl_als_string = str(ganzzahl)
    print(ganzzahl_als_string)
    anz_dezimalstellen = len(zahl1[len(ganzzahl_als_string) + 1:])
    dezimalstellen = zahl1[anz_dezimalstellen+1:]
    print(anz_dezimalstellen, dezimalstellen)

    zaehler = int(dezimalstellen)
    nenner = 10 ** anz_dezimalstellen

    from math import gcd
    kuerzer = gcd(zaehler, nenner)

    print(zaehler, nenner, kuerzer)

    zaehler = int(zaehler / kuerzer)
    nenner = int(nenner / kuerzer)

    return ganzzahl, zaehler, nenner


print(brueche('2.95'))
````
