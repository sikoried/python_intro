# Diamantenwort

Für ein eingegebenes Wort soll ein Diamant ausgegeben werden, welcher aus den Buchstaben des Wortes (in Wiederholung) gebildet wird.
Der Diamant ist an der breitesten Stelle so breit wie das Wort lang ist, und beginnt immer mit einem Großbuchstaben, gefolgt von Kleinbuchstaben.

# Beispiele:

Eingabe:
```
beispiel
```

Ausgabe:
```
   Be
  Ispi
 Elbeis
Pielbeis
 Pielbe
  Ispi
   El
```

Eingabe:
```
Peter
```

Ausgabe:
```
  P
 Ete
Rpete
 Rpe
  T
```


````python
wort1 = "Beispiel"
wort2 = "Peter"


def drucker(wort):
    n = len(wort)
    anz_leer = int((n-1)/2)
    anz_buchst = 2 - (n % 2)
    letzte = int(n / 2) - (1 - n % 2)
    print(n, anz_leer, anz_buchst, letzte)
    pos = 0
    for x in range(n-(1 - n % 2)):
        zeile = anz_leer * ' '
        wortbaustein = ''
        for i in range(0, anz_buchst):
            wortbaustein = wortbaustein + wort[(pos + i) % len(wort)]
        pos += anz_buchst

        print(zeile + wortbaustein.capitalize())
        if x < letzte:
            anz_leer = anz_leer - 1
            anz_buchst = anz_buchst + 2
        else:
            anz_leer = anz_leer + 1
            anz_buchst = anz_buchst - 2

drucker(wort1)
drucker(wort2)
````

Schreiben Sie ein Programm, dass eine Buchstabenkombination durch Aufsummierung der Buchstabenwerte in eine Zahl umwandelt.
Dabei ist `A` 1, `B` 10, `C` 100, usw. bis `G` 1000000 wert.

Es werden nun zeilenweise Kombination beliebiger Länge der Buchstaben A-G eingegeben, für die jeweils die Dezimalzahl ausgegeben werden soll.

# Beispiele:

Eingabe:
```
CCBBBBA
ABC
```

Ausgabe:
```
241
111
```


````python
ein1 = "CCBBBBA"
aus1 = 241

ein2 = "ABC"
aus1 = 111

def roemisch(w):
    sum = 0
    for b in w:
        if b == 'A':
            pot = 0
        elif b == 'B':
            pot = 1
        elif b == 'C':
            pot = 2
        elif b == 'D':
            pot = 3
        elif b == 'E':
            pot = 4
        elif b == 'F':
            pot = 5
        elif b == 'G':
            pot = 6
        else:
            raise Exception("Unbekannter Buchstabe")
        
        sum = sum + 10 ** pot
    
    return sum

def roemisch(w):
    sum = 0
    offset = int.from_bytes('A'.encode('ascii'))
    for b in w:
        pot = int.from_bytes(b.encode('ascii')) - offset
        sum += 10 ** pot
    return sum

roemisch(ein1)
````
