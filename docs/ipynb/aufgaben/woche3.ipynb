# Wortlängen

Schreiben Sie ein Programm, dass eine beliebige Anzahl Zeilen einliest und abschließend ausgibt welche Wortlängen wie oft vorgekommen sind; hierbei sind Wörter mit neun oder mehr Buchstaben als gleich lang zu betrachten.
Wenn die Anzahl von Wörtern mit einer bestimmten Länge 0 ist, wird sie nicht ausgegeben, Satzzeichen soll ignoriert werden.

# Beispiele:

Eingabe:
```
Hi Sam, how are you? I really wonder
what this is all about: Schiffahrtskapitän
```

Ausgabe:
```
1 1
2 2
3 5
4 2
5 1
6 2
9 1
```


````python
zeilen = [
    "Hi Sam, how are you? I really wonder",
    "what this is all about: Schiffahrtskapitän"
]

import re

laengen = dict()  # Laenge => Anzahl
for z in zeilen:
    for w in re.findall(r'\w+', z):
        l = min(len(w), 9)

        anzahl_vorher = laengen[l] if l in laengen else 0

        # Ausführlich:
        # if l in laengen:
        #     anzahl_vorher = laengen[l]
        # else:
        #     anzahl_vorher = 0

        # Dictionary-spezifisch mit Defaultwert
        # anzahl_vorher = laengen.get(l, 0)

        anzahl_nacher = anzahl_vorher + 1
        laengen[l] = anzahl_nacher

# for k in sorted(laengen):
#     print(k, laengen[k])

# print(list(sorted(laengen.items(), key=lambda x: x[1])))

for l, a in sorted(laengen.items()):
    print(l, a)
````

# Buchstabendreieck

Für einen eingegebenen Buchstaben soll das Alphabet ab dem Buchstaben A bis zu dem eingegebenen Buchstaben ausgegeben werden. 
Die Anzahl der Buchstaben pro Zeile entspricht der Zeilennummer.
Ist die Anzahl der verbleibenden Buchstaben kleiner als die nötigen Buchstaben pro Reihe wird mit `*` aufgefüllt.

Alle Buchstaben sind Großbuchstaben.

# Beispiele:

Eingabe:
```
D
```

Ausgabe:
```
A
BC
D**
```

````python
buchstabe = 'G'

offset = ord('A')  # 65
b = ord(buchstabe) - offset  # 3

alph = ""
for i in range(0, b+1):
    alph = alph + chr(offset+i)

i = 0
z = 1
while i < len(alph):
    # print(f"aktuell auszugeben: start={i} anzahl={z}")
    ausgabe = alph[i:i+z]

    bekommen = len(ausgabe)

    if bekommen < z:
        ausgabe += ('*' * (z-bekommen))

    print(ausgabe)

    i = i + z
    z = z + 1
````

# Noch alle Klammern im Schrank?

Überprüfen Sie für jede Zeile der Eingabe, ob die Anzahl der Klammern korrekt ist, d.h. dass alle öffnenden Klammern auch geschlossen werden, und geben Sie entsprechend `true` oder `false` aus.


# Beispiele

## Eingabe
```
(3+(7*2)-6)
)(
((hans dampf))
(hans)(peter))
```

## Ausgabe
```
true
false
true
false
```

````python
test1 = "(3+(7*2)-6)"
test2 = ")("
test3 = "((hans dampf))"
test4 = "(hans)(peter))"

def pruefeklammern(s):
    n = 0
    for i, z in enumerate(s):
        if z == '(':
            n += 1  # n = n + 1
        elif z == ')':
            n -= 1  # n = n - 1
        
        if n < 0:
            return "false", i

    if n > 0:
        return "false", i
    else:
        return "true", i

print(pruefeklammern(test1))
print(pruefeklammern(test2))
print(pruefeklammern(test3))
print(pruefeklammern(test4))
````
