### Daten von einer API verwenden

Das Projekt (OpenMensa)[https://openmensa.org] sammelt Informationen zu Standorten und Speiseplänen öffentlicher Mensen, und bietet diese als Javascript Object Notation (JSON) an.


```python
import requests

response = requests.get('https://openmensa.org/api/v2/canteens/')

if response.ok:
    payload = response.json()
else:
    print(f'Request was not successful and returned code: {response.status_code}.')
    payload = None

mensen = pd.DataFrame(payload)
mensen.head(3)
```

*Tip: `df.to_csv('data.csv')` schreibt diese Daten in `data.csv`.*
