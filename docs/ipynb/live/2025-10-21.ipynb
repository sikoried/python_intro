```python
def yfun(t, v0):
    g = 9.81
    y = v0*t - 0.5 * g * t ** 2
    dydt = v0 - g * t
    return y, dydt

position, velocity = yfun(10, 3)
print(position, velocity)
```

```python
from math import log

def lnapprox(x, tol=1e-4):
    """
    Dokumentation für lnapprox.
    """
    ideal = -log(1 - x)
    s = 0
    i = 1
    while abs(ideal-s) > tol:
        s = s + (x ** i) / i
        i = i + 1

    return ideal, s, i, abs(ideal-s)


def table():
    for x in range(2, 10):
        print(lnapprox(1/x))

table()
```

```python
def diff2(f, x, h=1e-6):
    r = (f(x-h) - 2 * f(x) + f(x+h))/(h**2)
    return r

def f(x):
    return x ** 2 - 1

for x in range(0, 10):
    print(diff2(lambda x: x ** 3 - 1, x))
```

```python
def bisection(f, a, b, tol=1e-3):
    if f(a) * f(b) > 0:
        print(f"Keine Nullstelle oder mehrere zwischen [{a}, {b}]")
        return
    m = (a+b)/2

    while abs(f(m)) > tol:
        if f(a) * f(m) < 0:
            b = m
        else:
            a = m
        m = (a+b)/2
    return m

def f(x):
    return x**2 - 1

print(bisection(f, -10, +10))
```

```python
def fib(n):
    if n == 1:
        return 1
    elif n == 2:
        return 1
    else:
        return fib(n-2) + fib(n-1)
    

for x in range(1, 10):
    print(x, fib(x))


fib(100)
```
