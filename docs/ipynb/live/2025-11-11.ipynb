```python
import requests

response = requests.get('https://openmensa.org/api/v2/canteens/')

if response.ok:
    payload = response.json()
else:
    print(f'Request was not successful and returned code: {response.status_code}.')
    payload = None

mensen = pd.DataFrame(payload)
mensen.head(3)
```

```python
response = requests.get('https://openmensa.org/api/v2/canteens/268/meals/')

if response.ok:
    payload = response.json()
else:
    print(f'Request was not successful and returned code: {response.status_code}.')
    payload = None

meals = pd.DataFrame(payload)
meals.head(3)

meals = meals.explode('meals')
# meals['meals']
meals[meals['meals'].apply(lambda x: x['category'] in ['Vegan', 'Vegetarisch'])]
```

```python
zeilen = [
    "Mary had a little lamb.",
    "Peter paid a pretty penny.",
    "Java sucks."
]

woerter = []
bisher_laengstes = 0
for z in zeilen:
    z = z.split()
    if len(z) > bisher_laengstes:
        bisher_laengstes = len(z)
    woerter.append(z)

for i in range(0, bisher_laengstes):
    hilf = ''
    for w in woerter:
        hilf = hilf + (w[i] if i < len(w) else '-') + ' '
    print(hilf)
```

```python
# 1-2-hop
teiler = 3
anzahl = 16


for x in range(1, anzahl+1):
    print('hop' if x % teiler == 0 else x)

x = 1
while x <= anzahl:
    print('hop' if x % teiler == 0 else x)
    x = x + 1
```
