```python
class Line:
    def __init__(self, c0, c1):
        self.c0 = c0
        self.c1 = c1
    
    def __call__(self, x):
        return self.c0 + self.c1 * x
    
    def table(self, start, end, step):
        punkte = []
        for x in range(start, end, step):
            punkte.append((x, self(x)))
        return punkte
    

l1 = Line(0, 1)
l2 = Line(1, 2)

for t in l2.table(-5, 5, 1):
    print(t)
```

```python
class Parabola(Line):
    def __init__(self, c0, c1, c2):
        super().__init__(c0, c1)
        self.c2 = c2
    
    def __call__(self, x):
        return super().__call__(x) + self.c2 * x**2
    

p1 = Parabola(0, 0, 1)
for t in p1.table(-5, 5, 1):
    print(t)
```

```python
code = 0
text = "Peter Pan ist ein Held."
chiff0 = "Peter Pan ist ein Held."
chiff2 = "Rgvgt Rcp kuv gkp Jgnf."

def caesar(c, zeile):
    chiff = ""

    offA = ord('A')
    offZ = ord('Z')
    offa = ord('a')
    offz = ord('z')

    for b in zeile:
        b = ord(b)
        if b >= offA and b <= offZ:
            b = chr((((b - offA) + c) % 24) + offA)
        elif b >= offa and b <= offz:
            b = chr((((b - offa) + c) % 24) + offa)
        else:
            b = chr(b)
        chiff = chiff + b
    
    return chiff

caesar(-2, text)
```
