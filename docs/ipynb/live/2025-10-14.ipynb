# Loops und Listen

```python
K = 100
r = 2

n = 1; E = K * (1 + (r/100))**n; print(f"Kontostand nach {n} Jahren: {E:.2f}")
n = 2; E = K * (1 + (r/100))**n; print(f"Kontostand nach {n} Jahren: {E:.2f}")
n = 3; E = K * (1 + (r/100))**n; print(f"Kontostand nach {n} Jahren: {E:.2f}")
n = 4; E = K * (1 + (r/100))**n; print(f"Kontostand nach {n} Jahren: {E:.2f}")
n = 5; E = K * (1 + (r/100))**n; print(f"Kontostand nach {n} Jahren: {E:.2f}")
```

```python
n = 1
E = 0
while E < 200:
    E = K * (1 + (r/100))**n
    print(f"Kontostand nach {n} Jahren: {E:.2f}")
    n = n + 5

print("Fertig")
```

```python
zahl = "11001"
zahl = "010"

ergebnis = (zahl[0] == '1')

for buchstabe in zahl:
    ergebnis = (ergebnis and (buchstabe == '1' or buchstabe == '0'))

print(ergebnis)
```

```python
zahl = "1010"

e = 0
erg = 0
for z in reversed(zahl):
    erg = erg + int(z) * 2 ** e
    e = e + 1


print(erg)
```

```python
from math import sqrt

paare_von_triplets = [
    ((1, 2, 3), (2, 2, 1)),
    ((0, 0, 1), (0, 1, 1)),
    ((10, 10, 3), (10, 6, 1))
]

for (p1, p2) in paare_von_triplets:
    print(p1)
    print(p2)
    v = (p1[0]-p2[0], p1[1]-p2[1])
    lv = sqrt(v[0]**2 + v[1]**2)

    test = lv - p1[2] - p2[2]

    beruehren = (test == 0)
    schneiden = (test < 0)

    print(f"Berührung: {beruehren}")
    print(f"Schneiden: {schneiden}")
```
