```python
p = {0: -1, 2: 1, 7: 3}

def poly(poly, x):
    sum = 0.0
    for potenz in poly:
        sum = sum + poly[potenz] * (x ** potenz)
    return sum

poly(p, 4)
```

```python
p = {0: -1, 2: 1, 7: 3}

def poly(poly, x):
    sum = 0.0
    for (potenz, faktor) in poly.items():
        sum = sum + faktor * (x ** potenz)
    return sum

poly(p, 4)
```

# Übungsaufgaben

```python
satz = "All is quiet now, but wait!"
buchstaben = "QAET"

# q 1
# a 2
# e 1
# t 3

zaehler = {}  # dict()

# dictionary befüllen
for b in buchstaben.lower():
    zaehler[b] = 0

# buchstaben zählen
for b in satz.lower():
    if b in zaehler:
        zaehler[b] = zaehler[b] + 1

# dictionary nach keys ausgeben
for k in zaehler:
    print(k, zaehler[k])
```
